```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Class labels
classes = ['DCM', 'HCM', 'MINF', 'NOR', 'RV']

# Confusion matrices
cm_tsetlin = np.array(
    [[7 2 1 0 0]
 [0 3 1 6 0]
 [2 3 5 0 0]
 [0 3 0 4 3]
 [0 0 0 2 8]])

cm_dnn = np.array([
    [9, 1, 0, 0, 0],
    [1, 6, 0, 3, 0],
    [3, 3, 4, 0, 0],
    [0, 3, 0, 5, 2],
    [0, 1, 0, 2, 7]
])

# Normalize each row
cm_tsetlin_norm = cm_tsetlin / cm_tsetlin.sum(axis=1, keepdims=True)
cm_dnn_norm = cm_dnn / cm_dnn.sum(axis=1, keepdims=True)

# Plot
fig, axes = plt.subplots(1, 2, figsize=(11, 4.8))

sns.heatmap(
    cm_tsetlin_norm,
    annot=True,
    fmt='.2f',
    cmap='Blues',
    vmin=0,
    vmax=1,
    cbar=False,
    xticklabels=classes,
    yticklabels=classes,
    ax=axes[0]
)

axes[0].set_xlabel('Predicted')
axes[0].set_ylabel('True')
axes[0].set_title('(a) Tsetlin')

sns.heatmap(
    cm_dnn_norm,
    annot=True,
    fmt='.2f',
    cmap='Blues',
    vmin=0,
    vmax=1,
    cbar=False,
    xticklabels=classes,
    yticklabels=classes,
    ax=axes[1]
)

axes[1].set_xlabel('Predicted')
axes[1].set_ylabel('True')
axes[1].set_title('(b) DNN')

plt.tight_layout()

# Save for Overleaf
plt.savefig(
    'confusion_matrices.pdf',
    bbox_inches='tight',
    dpi=300
)

plt.show()
```
